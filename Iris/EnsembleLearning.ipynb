```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
X = df.drop('species',axis=1)
y = df['species']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.ensemble import StackingClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression

base_learners = [
    ('svc',SVC()),
    ('dt', DecisionTreeClassifier())
]
meta_learner = LogisticRegression(max_iter=10)

stacking_classifier = StackingClassifier(estimators = base_learners, final_estimator = meta_learner, cv = 5)
stacking_classifier.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report

y_pred = stacking_classifier.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
accuracy
```

```python
from sklearn.ensemble import RandomForestClassifier
random_forest_classifier = RandomForestClassifier(n_estimators=110,max_depth=2)
random_forest_classifier.fit(X_train, y_train)
```

```python
y_pred = random_forest_classifier.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
accuracy
```

```python
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier

adaboost_classifier = AdaBoostClassifier(n_estimators=10)
adaboost_classifier.fit(X_train, y_train)
```

```python
y_pred = adaboost_classifier.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
accuracy
```

```python
gradient_classifier = GradientBoostingClassifier(learning_rate=0.1)
gradient_classifier.fit(X_train, y_train)
y_pred = gradient_classifier.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
accuracy
```
