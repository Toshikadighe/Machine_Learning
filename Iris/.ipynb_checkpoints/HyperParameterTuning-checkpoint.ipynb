```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df['species'].unique()
```

```python
df.shape
```

```python
X = df.drop('species',axis=1)
y = df['species']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
model_knn = KNeighborsClassifier(n_neighbors=3)
model_knn.fit(X_train, y_train)
model_knn.score(X_test,y_test)
```

```python
from sklearn.svm import SVC
model_svc = SVC(gamma='auto', C=10)
model_svc.fit(X_train, y_train)
model_svc.score(X_test,y_test)
```

```python
from sklearn.model_selection import GridSearchCV
param_grid = {'C' : [0.1,1,10,20],
              'gamma' : ['auto', 'scale'],
              'kernel' : ['rbf', 'linear']
             }
model_grid_search = GridSearchCV(model_svc, param_grid, cv =5)
model_grid_search.fit(X_train, y_train)
```

```python
model_grid_search.cv_results_
```

```python
results = pd.DataFrame(model_grid_search.cv_results_)
results
```

```python
results[['params', 'mean_test_score']]
```

```python
param_grid = {'n_neighbors' : [3,5,7,9,11,13],
              'metric' : ['euclidean','manhattan']
             }
model_grid_search = GridSearchCV(model_knn, param_grid, cv =5)
model_grid_search.fit(X_train, y_train)
results = pd.DataFrame(model_grid_search.cv_results_)
results[['params', 'mean_test_score']]
```

```python
from sklearn.model_selection import RandomizedSearchCV
param_grid = {'C' : [0.1,1,10,20],
              'gamma' : ['auto', 'scale'],
              'kernel' : ['rbf', 'linear']
             }
model_random_search = RandomizedSearchCV(model_svc, param_grid, cv =5, n_iter = 4)
model_random_search.fit(X_train, y_train)
results = pd.DataFrame(model_random_search.cv_results_)
results[['params', 'mean_test_score']]
```

```python
param_grid = {'n_neighbors' : [3,5,7,9,11,13],
              'metric' : ['euclidean','manhattan']
             }
model_random_search = RandomizedSearchCV(model_knn, param_grid, cv =5, n_iter = 4)
model_random_search.fit(X_train, y_train)
results = pd.DataFrame(model_random_search.cv_results_)
results[['params', 'mean_test_score']]
```
