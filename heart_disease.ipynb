```python
df_clean = pd.get_dummies(df,drop_first=True)
df_clean
```

```python
bool_types = df_clean.select_dtypes(include='bool')
bool_types.columns
```

```python
df_clean[bool_types.columns] = df_clean[bool_types.columns].astype(int)
df_clean
```

```python
from sklearn.preprocessing import StandardScaler
scale_cols=['Age','RestingBP','Cholesterol','MaxHR','Oldpeak']
scaler=StandardScaler()
df_clean[scale_cols] = scaler.fit_transform(df_clean[scale_cols])
df_clean.head()
```

```python
X = df_clean.drop('HeartDisease',axis=1)
y = df_clean['HeartDisease']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model_logistic = LogisticRegression()
model_logistic.fit(X_train,y_train)
```

```python
y_pred_logistic = model_logistic.predict(X_test)
y_pred_logistic
```

```python
from sklearn.metrics import accuracy_score,classification_report
accuracy = accuracy_score(y_test, y_pred_logistic)
accuracy
```

```python
print(classification_report(y_test, y_pred_logistic))
```
