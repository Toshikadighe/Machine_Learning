```python
df['bedrooms'].unique()
```

```python
df[df['bedrooms'] == 0]
```

```python
(df['bathrooms'] == 0).sum()
```

```python
df['bedrooms']=df['bedrooms'].replace(0,np.nan)
df['bathrooms']=df['bathrooms'].replace(0,np.nan)
```

```python
df['bedrooms'].fillna(df['bedrooms'].median(),inplace=True)
df['bathrooms'].fillna(df['bathrooms'].median(),inplace=True)
```

```python
(df['bedrooms'] == 0).sum()
```

```python
df['floors'].unique()
```

```python
df['was_renovated'] = (df['yr_renovated']>0).astype(int)
```

```python
df['last_updated'] = df['yr_renovated'].where(df['yr_renovated']>0,df['yr_built'])
df['years_since_update'] = 2015 - df['last_updated']
```

```python
df.drop('yr_renovated',axis=1,inplace=True)
df.head()
```

```python
df['date'].unique()
```

```python
df = df.drop(columns=['date'])
```

```python
df['country'].unique()
```

```python
df = df.drop(columns=['country'])
```

```python
df['statezip'].str.contains('WA').sum()
```

```python
df['zipcode'] = df['statezip'].str.split(' ').str[1]
```

```python
df = df.drop(columns=['statezip'])
df.head()
```

```python
df['street'].nunique()
```

```python
df = df.drop(columns=['street'])
df['city'].nunique()
```

```python
df['zipcode'].nunique()
```

```python
df['city'].value_counts()
```

```python
df['zipcode'].value_counts()
```

```python
df = df.drop(columns=['zipcode'])
```

```python
df = pd.get_dummies(df, columns=['city'], drop_first=True, dtype=int) 
df.head()
```

```python
for col in ['sqft_living', 'sqft_lot']:
    df[col + '_log'] = np.log1p(df[col])
```

```python
(df['price'] == 0).sum()
```

```python
df = df[df['price'] != 0]
```

```python
df['price_log'] = np.log1p(df['price'])      
sns.histplot(df['price_log'], kde=True)
```

```python
numeric_cols = ['price_log','sqft_living_log', 'sqft_lot_log']
for col in numeric_cols:
    plt.figure(figsize=(4,6))
    sns.histplot(data=df[col],kde=True)
```

```python
df_clean = df.drop(columns=['price','sqft_living', 'sqft_lot', 'sqft_above','sqft_basement'])
df_clean.head()
```

```python
df_clean.corr(numeric_only=True)['price_log'].sort_values(ascending=False)
```

```python
df_clean.drop(columns=['last_updated','years_since_update'],inplace=True)
```

```python
df_clean.drop(columns=['sqft_lot_log'],inplace=True)
df_clean.columns
```

```python
X = df_clean.drop('price_log',axis=1)
y = df['price_log']
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
rf.fit(X, y)

importances = pd.Series(rf.feature_importances_, index=X.columns)
importances.sort_values(ascending=False).head(20)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
X_train.columns
```

```python
X_train['view'].value_counts()
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
cols = ['bedrooms', 'bathrooms', 'floors', 'view', 'condition', 'yr_built',
       'sqft_living_log']
X_train[cols]=scaler.fit_transform(X_train[cols])
```

```python
X_train.head()
```

```python
X_test[cols]=scaler.transform(X_test[cols])
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
pred_dollars = np.expm1(y_pred)
actual_dollars = np.expm1(y_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error

print("MAE ($):", mean_absolute_error(actual_dollars, pred_dollars))
print("R²:", r2_score(actual_dollars, pred_dollars))
```

```python
from sklearn.linear_model import RidgeCV
alphas = np.logspace(-3, 3, 50)   
model = RidgeCV(alphas=alphas, cv=5)
model.fit(X_train, y_train)

print("Best alpha:", model.alpha_)
```

```python
from sklearn.linear_model import Ridge
model_ridge = Ridge(alpha=0.15)
model_ridge.fit(X_train, y_train)
```

```python
y_pred = model_ridge.predict(X_test)
y_pred
```

```python
pred_dollars = np.expm1(y_pred)
actual_dollars = np.expm1(y_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error

print("MAE ($):", mean_absolute_error(actual_dollars, pred_dollars))
print("R²:", r2_score(actual_dollars, pred_dollars))
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error

rf = RandomForestRegressor(n_estimators=200, n_jobs=-1)
rf.fit(X_train, y_train)
pred = np.expm1(rf.predict(X_test))
print("RF R²:", r2_score(actual_dollars, pred))
print("RF MAE:", mean_absolute_error(actual_dollars, pred))
```
